```python
import torch
from torch import nn
import matplotlib.pyplot as plt
torch.__version__
```

linear rgration formula ---> [Y = a + bX]

b = slop(gradient)

weight = b

bias = a

neuron = X

b1X1 + b2X2 + b3X3 + a = Y


```python
weight = 0.9
bias = 0.1

start = 0
end = 1
step = 0.01
X = torch.arange(start, end, step).unsqueeze(dim=1)
Y = bias + weight*X

split_index = int(0.7 * len(X))
X_train, Y_train = X[:split_index], Y[:split_index]
X_test, Y_test = X[split_index:], Y[split_index:]
len(X_train), len(Y_train), len(X_test), len(Y_test)
```

```python
def plot_predictions(train_data = X_train,
                     train_result_labels = Y_train,
                     test_data = X_test,
                     test_result_labels = Y_test,
                     predictions = None ):
  plt.figure(figsize = (15, 10))

  plt.scatter(train_data, train_result_labels, s=3, c="red",label = "training data")

  plt.scatter(test_data, test_result_labels, s=3, c="green",label = "testing data")

  if predictions is not None:
      plt.scatter(test_data, predictions, s=3, c="black",label = "predictions") # Corrected to plot 'predictions'

  plt.legend(prop = {"size": 20}); # Moved inside function
  plt.show(); # Moved inside function
```

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(1,dtype=torch.float),requires_grad=True)

        self.bias = nn.Parameter(torch.randn(1,dtype=torch.float), requires_grad=True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.weights * x + self.bias
```

```python

torch.manual_seed(35)
model_0 = LinearRegressionModel()

list(model_0.parameters())
```

```python
model_0.state_dict
```

```python
with torch.inference_mode():
  y_predictions = model_0(x=X_test)
plot_predictions(predictions=y_predictions)
```

  TRAINING TIME !!

```python
loss_function = nn.L1Loss()

optimizer = torch.optim.SGD(params=model_0.parameters(),lr=0.001)

```

```python
epochs = 1000

for epoch in range (epochs):
  model_0.train()

  y_predictions = model_0(X_train)

  loss = loss_function(y_predictions, Y_train)

  optimizer.zero_grad()

  loss.backward()

  optimizer.step()

```
